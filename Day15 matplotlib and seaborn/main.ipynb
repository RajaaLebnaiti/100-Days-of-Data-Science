## Explication:

#### Histogrammes : simple, density, cumulative

---

#### ⚡ Étape 0 : Données
On prend les valeurs suivantes :  

\[
data = [1, 2, 2, 3, 4, 5, 20, 20]
\]

---

####  ⚡ Étape 1 : Histogramme normal (sans density, sans cumulative)

- Nombre de bins = 4  
- Min = 1, Max = 20  
- Largeur de bin = \((20 - 1)/4 = 4.75\)  

Donc les intervalles sont :  

- Bin 1 : [1.0, 5.75) → {1,2,2,3,4,5} → **6 valeurs**  
- Bin 2 : [5.75, 10.5) → {} → **0 valeur**  
- Bin 3 : [10.5, 15.25) → {} → **0 valeur**  
- Bin 4 : [15.25, 20.0] → {20,20} → **2 valeurs**  

👉 Effectifs = **[6, 0, 0, 2]**

---

####  ⚡ Étape 2 : Histogramme avec `density=True`

Formule de la hauteur d’un bin :  

\[
h_i = \frac{effectifs_i}{N \times \text{largeur du bin}}
\]

Avec :  

- \(N = 8\)  
- largeur du bin = 4.75  

Calcul :  

- Bin 1 : \(6 / (8 \times 4.75) \approx 0.158\)  
- Bin 2 : 0  
- Bin 3 : 0  
- Bin 4 : \(2 / (8 \times 4.75) \approx 0.0526\)  

👉 Vérification aire totale :  

\[
0.158 \times 4.75 + 0.0526 \times 4.75 = 0.75 + 0.25 = 1.0 \; \checkmark
\]

---

#### ⚡ Étape 3 : Histogramme avec `cumulative=True`

On cumule les effectifs :  

- Bin 1 = 6  
- Bin 2 = 6  
- Bin 3 = 6  
- Bin 4 = 8  

👉 Avec `density=True + cumulative=True`, on cumule les **probabilités** :  

- Bin 1 = 6/8 = 0.75  
- Bin 2 = 0.75  
- Bin 3 = 0.75  
- Bin 4 = 1.0  


```python

data = np.random.randn(1000)  # 1000 valeurs aléatoires (moyenne 0, écart-type 1)

plt.figure(figsize=(10, 6))

# Histogramme simple
plt.subplot(2, 2, 1)
plt.hist(data, bins=10, color='skyblue', edgecolor='black')
plt.title("Histogramme simple (10 bins)")

# Histogramme avec plus de détails
plt.subplot(2, 2, 2)
plt.hist(data, bins=30, color='orange', edgecolor='black')
plt.title("Histogramme détaillé (30 bins)")

# Histogramme densité
plt.subplot(2, 2, 3)
plt.hist(data, bins=20, density=True, color='green', alpha=0.7, edgecolor='black')
plt.title("Histogramme en densité")

# Histogramme cumulé
plt.subplot(2, 2, 4)
plt.hist(data, bins=20, cumulative=True, color='red', alpha=0.7, edgecolor='black')
plt.title("Histogramme cumulé")

plt.tight_layout()
plt.show()
```

## Différence entre scatter() et plot()

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
x = np.linspace(0, 10, 30)
y = np.sin(x) + 0.2*np.random.randn(len(x))

plt.figure(figsize=(10,4))

# subplot A : scatter (pas de lignes) avec tailles et couleurs différentes
plt.subplot(1,2,1)
sizes = (np.linspace(10, 200, len(x)))   # tailles croissantes
colors = y                               # couleur selon y
sc = plt.scatter(x, y, s=sizes, c=colors, cmap='coolwarm', alpha=0.8, edgecolors='k')
plt.colorbar(sc, label='valeur y (colormap)')
plt.title("scatter : points indépendants")
plt.xlabel("x")
plt.ylabel("y")

# subplot B : plot (points reliés) — on montre ligne + marqueurs
plt.subplot(1,2,2)
plt.plot(x, y, '-o', linewidth=2, markersize=6, label='ligne + points')
plt.title("plot : points reliés par une ligne")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()

plt.tight_layout()
plt.show()
```

#### b)- sns.displot()


```python
import seaborn as sns
import matplotlib.pyplot as plt
tips = sns.load_dataset('tips')
```

```python
sns.displot(data=tips, x="total_bill", hue="sex", kind="hist", multiple="stack")
plt.show()
```

```python
sns.displot(data=tips, 
            x="total_bill", 
            col="time", 
            row="sex", 
            bins=20,
            kde= True)

plt.show()
```

#### c)- lmplot()

```python
iris = sns.load_dataset('iris')
iris.head(3)
```

```python
sns.lmplot(data=iris, x="sepal_length", y="sepal_width", hue='species')
plt.show()
```

```python
sns.lmplot(data=iris, x="petal_length", y="petal_width", hue='species')
plt.show()
```

```python
sns.lmplot(
    data=tips,
    x="total_bill",
    y="tip",
    col="time",   # séparer Lunch vs Dinner
    row="sex",    # séparer Male vs Female
    hue="smoker"  # comparer Fumeur vs Non-fumeur
)
plt.show()
```

#### d)- catplot()

```python
sns.catplot(data=tips, x="day", y="total_bill")
plt.show()
```
