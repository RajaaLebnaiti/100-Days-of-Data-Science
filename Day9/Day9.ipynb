#  Filtering

<h3> Logical operators

```python
# student older than 17
print(data [data['age']> 17] )
```

```python
# girls with score > 85
# USE ( ) to sepereate between each condition !!!!!!!!!!
print(data[ (data['gender']=='F') & (data['score']> 85) ])
```

<h3>Using .filter()

<h4> dataframe.filter(items, like, regex, axis)


```python
print(data.filter(items=["name","age"]))
```

<h3> Using isin()

```python
names = ['John','Catlin','Mike', 'Alice']
print(data[data.name.isin(names)])
```

# Using str accessor

```python
data[data.name.str.startswith('D')]
```

```python
data[data.name.str.contains('a')]
```

# Using ~ (not )

```python
data[~data.name.str.contains('o')]
```

## Using query

```python
data.query('  age > 16 and   gender == "F" ')
```

## Using loc and iloc

```python
import pandas as pd
data = pd.read_csv('data.csv')
data.head()
```

- Loc: Select rows or columns using labels


- Iloc: Select rows or columns using indices

```python
data.iloc[3:5, :] # rows 3 and 4 all columns
```

```python
data.loc[3:5, :] # rows 3 and 4 all columns
```

We can see the same results.

# Grouping

```python
# average score by class

# we group by class # Then for each class we calculate the mean of scores
print(data.groupby("class")["score"].mean())
```

```python
# # Max score by gender

print(data.groupby("gender")["score"].max())
```

# Handling missing values

```python
data.isnull().sum()
```

```python
# # Fill missing ages with the mean

data["age"].fillna(data["age"].mean(), inplace= True)
```

```python
# # Fill missing scores with 0

data["score"].fillna(0, inplace= True)
```

```python
data.isnull().sum()
```

# Merging

```python
#  Create a second table with ID and email
df_emails = pd.DataFrame({
    "id": [1, 2, 3, 4, 5, 6],
    "email": ["a@x.com", "b@x.com", "c@x.com", "d@x.com", "e@x.com", "f@x.com"]
})
```

```python
df_emails
```

```python
data
```

```python
# merging based on id

data_merged= data.merge(df_emails, on="id")
data_merged
```

```python
merged = pd.merge(data, df_emails, on="id")
print(merged.head())
```
