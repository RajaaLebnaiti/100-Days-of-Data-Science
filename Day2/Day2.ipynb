```python
# Write a function to return the longest string from a list of strings.

def find_longest_string(strings):
    longest= ""
    for i in strings:
        if len(i)> len(longest):
            longest= i
    return longest


list_of_strings= ["Hello", "Rajaa", "a", "abdcffff"]    
find_longest_string(list_of_strings)
```

```python
#  Filter List of Integers Greater than 5
def greater_than_5(integers):
    liste= []
    for i in integers:
        if i> 5:
            liste.append(i)
    return liste


liste = [1,2,3,3,5,6,7,8,9,10]
greater_than_5(liste)
```
