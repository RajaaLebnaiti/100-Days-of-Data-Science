# 1)- Label Encoder / One Hot Encoder

```python
import pandas as pd 
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
```

```python
data = pd.DataFrame({
    "couleur": ['rouge', 'vert', 'bleu', 'orange'],
    "taille": ['petit', 'grand', 'moyen', 'petit']
})
data
```

```python
# label encoder pour les variables ordinales
le = LabelEncoder()
data['taille_le'] = le.fit_transform(data['taille'])


data
```

```python
# one hot encoder pour les variables nominales


ohe = OneHotEncoder(sparse_output=False)  
encoded = ohe.fit_transform(data[["couleur"]])

# Convertir en DataFrame
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out(["couleur"]))

# Concaténer avec dataset
data_final = pd.concat([data, encoded_df], axis=1)


data_final
```

```python
# appliquer sur titanic dataset

import seaborn as sns

titanic= sns.load_dataset('titanic')
titanic.dropna(inplace=True)
titanic.drop_duplicates(inplace=True)
# pour faciliter
titanic.drop(columns=['who','pclass','adult_male', 'deck', 'sibsp',	'parch'	,'fare'	,'embarked', 'alone', 'alive'], inplace=True)
titanic.head()
```

```python
# label encoder to class

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
titanic['class_le'] = le.fit_transform(titanic['class'])
titanic.head()
```

```python
# one hot encoder pour les colonnes sex et cherboug
from sklearn.preprocessing import OneHotEncoder
ohe = OneHotEncoder(sparse_output= False)
encoded = ohe.fit_transform(titanic[['sex', 'embark_town']])
encoded_df = pd.DataFrame(encoded, columns= ohe.get_feature_names_out(['sex', 'embark_town']))
data_final = pd.concat([titanic, encoded_df], axis=1)

data_final.head()
```

## Différence entre label encoder et ordinal encoder


```python
data= pd.DataFrame({"taille": ['first', 'second', 'third', 'fourth']})

le = LabelEncoder()
data['taille_le']= le.fit_transform(data['taille'])

data # suit l'ordre alphabétique
```

```python
from sklearn.preprocessing import OrdinalEncoder

data= pd.DataFrame({"taille": ['first', 'second', 'third', 'fourth']})

oe = OrdinalEncoder(categories=[['first', 'second', 'third','fourth']])
data['taille_oe'] = oe.fit_transform(data[['taille']])
data
```
