<h1><center> Mise à l'échelle (Scaling)

<h2> a)- Standard Scaler

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
data = pd.DataFrame({"taille": [1,2,3,4,5]})
data
```

```python
print("avant standard scaler : ")
print(" ", data.mean(), "\n",np.sqrt(data.var()))
```

```python
ss = StandardScaler()
data['standard scaler'] = ss.fit_transform(data)
data
```

```python
print("après standard scaler : ")
print(" ", data['standard scaler'].mean(), "\n",data['standard scaler'].std())
```

<h2> b)- MinMaxScaler

```python
print("avant min max scaler: ")
print("min= ", data['taille'].min(), "max= ", data['taille'].max())
```

```python
from sklearn.preprocessing import MinMaxScaler
mms = MinMaxScaler()
data['minmaxscaler'] = mms.fit_transform(data[['taille']])

data
```

```python
print("après min max scaler: ")
print("min= ", data['minmaxscaler'].min(), "max= ", data['minmaxscaler'].max())
```

<h2> c)- RobustScaler

```python
from sklearn.preprocessing import RobustScaler

rs = RobustScaler()
data['robust scaler'] = rs.fit_transform(data[['taille']])
```

### Visualisations

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(12,6))

plt.subplot(311)
sns.boxplot(data= data['standard scaler'], orient='h')
plt.title("standard scaler")

plt.subplot(312)
sns.boxplot(data= data['minmaxscaler'], orient='h')
plt.title("minmax scaler")

plt.subplot(313)
sns.boxplot(data= data['robust scaler'], orient='h')
plt.title("ROBUST scaler")

plt.xlabel("TAILLE")
plt.ylabel("y values")

plt.show()
```
