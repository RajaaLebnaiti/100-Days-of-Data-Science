```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.preprocessing import Normalizer

# Exemple : deux features avec des vecteurs de tailles différentes
X = np.array([
    [3, 4],
    [1, 2],
    [5, 12],
    [8, 15]
])

# Normalisation L2
norm = Normalizer(norm="l2")
X_l2 = norm.fit_transform(X)

# Plot avant/après
fig, axes = plt.subplots(1, 2, figsize=(12, 6))

# Avant normalisation
axes[0].scatter(X[:, 0], X[:, 1], color="blue")
for i, (x, y) in enumerate(X):
    axes[0].arrow(0, 0, x, y, head_width=0.5, head_length=0.5, fc='blue', ec='blue')
axes[0].set_title("Avant normalisation (vecteurs de tailles différentes)")
axes[0].set_xlim(-1, 16)
axes[0].set_ylim(-1, 16)
axes[0].set_aspect('equal')

# Après normalisation L2
axes[1].scatter(X_l2[:, 0], X_l2[:, 1], color="red")
for i, (x, y) in enumerate(X_l2):
    axes[1].arrow(0, 0, x, y, head_width=0.05, head_length=0.05, fc='red', ec='red')
# Cercle unité
circle = plt.Circle((0,0), 1, color="gray", fill=False, linestyle="--")
axes[1].add_artist(circle)
axes[1].set_title("Après normalisation L2 (vecteurs ramenés sur le cercle unité)")
axes[1].set_xlim(-1.5, 1.5)
axes[1].set_ylim(-1.5, 1.5)
axes[1].set_aspect('equal')

plt.show()
```
