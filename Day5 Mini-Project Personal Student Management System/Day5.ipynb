2. Add Student
Ask user for:

- Name, age, grade, courses (as comma-separated string)

- Store as a dictionary inside a list

- Store courses as a set

Use function with *args for courses, **kwargs for extra info (like city or email)

```python
# Function to add a student

students = {}
def add_Student(*args, **kwargs):
    name = args[0]
    age = args[1]
    grade = args[2]
    courses = kwargs.get("courses", [])
    students.update({
        name:{
            "age": age, 
            "grade": grade, 
            "courses": set(courses)
            }
        }
    )
    
    print("The student has been added succesfully! The student info are:\n",students[name])
    
```

3. Search Student

    - Show number of unique courses using set

```python
# Function so search for a student by name

def search_student(name_stud):
    if name_stud in students.keys():
        print("Name: ",name_stud, "Age: ", students[name_stud]["age"], "grade: ", students[name_stud]["grade"], "courses: ", students[name_stud]["courses"])
    else:
        print("The student ",name_stud , "is not found! Try again.")
        
```

6. Save / Load Data to File
- Save student list to a file (.txt or .json)

```python
# save the list of students as json
import json 
def save_student():
    with open("student.json", "a") as file:
        json.dump(students, file)
    
    
# show the list of students
def display_students():
    with open("student.json", "r") as file:
        loaded = json.load(file)
        print("Before we begin, here is the list of students available for now: ", loaded)
```
