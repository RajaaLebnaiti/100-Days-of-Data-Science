```python
titanic.duplicated().sum()
```

```python
titanic.drop_duplicates(inplace=True)
titanic.duplicated().sum()
```

```python
# au lieu de supprimer les valeurs nulles, on les remplaces par mean, median, mode, ffill, bfill

data = sns.load_dataset('titanic')

data['age_mean'] = data['age'].fillna(data['age'].mean())
data['age_median'] = data['age'].fillna(data['age'].median())
data['age_mode'] = data['age'].fillna(data['age'].mode()[0])
data['age_ffill'] = data['age'].fillna(method="ffill")
data['age_bfill'] = data['age'].fillna(method="bfill")

print(data.iloc[888])  # one of the null values for age
```

```python
# gestion des valeurs bruitées


# Majuscule/Miniscule
df= pd.DataFrame({
    'sex' : ['male  ', "    Male", "MALE", "MaLe"]
})
print(df)
```

```python
df['sex']= df['sex'].str.lower()
print(df)
```

```python
df['sex']= df['sex'].str.strip() # supprime les espaces
print(df)
```

```python
# fautes de frappes

df= pd.DataFrame({
    'city': ['Paris', 'Pariss', 'pari']
})
print(df)
```

```python
df['city']= df['city'].replace({
    "pari": "Paris", 
    "Pariss": "Paris"})

print(df)
```

```python
# valuers numériques non logiques
# age= -5 age= 2000

df = pd.DataFrame({
                        "age": [1, 12, 23, -5, 2000, 20, 36]
})

print(df)
```

```python
df['age'] = df['age'].apply(lambda x: x if 0<= x <= 120 else np.nan)

print(df)
```

```python
# datetime formatée
df = pd.DataFrame({
                    "Date": ["2025-09-19", "19/09/25", "Sept 19 2025"]
})

print(df)
```

```python
df['Date'] = pd.to_datetime(df['Date'], errors="coerce")
print(df)
#Cela veut dire : si pandas n’arrive pas à convertir une valeur en date valide, il met NaT (Not a Time) au lieu de planter une erreur.
```

```python
# datetime formatée
df = pd.DataFrame({
                    "Date": ["hhhhh", "19/09/25", "Sept 19 2025"]
})

print(df)

df['Date'] = pd.to_datetime(df['Date'], errors="raise")
print(df)
```
