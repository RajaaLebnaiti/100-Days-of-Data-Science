<h4><b>Informations about the dataset</b></h4>

<ul>
    <li>Passenger ID</li>
    <li>Survived: 0 = No     1 = Yes</li>
    <li>Pclass: Ticket classe 1= 1st class, 2= 2nd class, 3= 3rd class </li>
    <li>Name: of passenger 891 unique values</li>
    <li>Sex: male 65% or female 35%</li>
    <li>Age: in years</li>
    <li>SibSp: Number of siblings/spouses abroad the titanic</li>
    <li>Parch: Number of parents/children abroad the titanic</li>
    <li>Ticket: Ticket number </li>
    <li>Fare: Passenger fare the price the passenger payd to travel on titanic</li>
    <li>Cabin: Cabin number</li>
    <li>Embarked: Port of Embarkation:C = Cherbourg, Q = Queenstown, S = Southampton</li>
