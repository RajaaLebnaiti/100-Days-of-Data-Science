```python
# duplicated values
df.duplicated().sum()
```

```python
# drop duplicates
df.drop_duplicates(inplace= True)
df
```

```python
# drop not usefull column
df.drop(columns= ['Not_Useful_Column'], inplace= True)
df
```

```python
# drop null values
df.dropna(inplace= True)
df
```

```python
df.isna().sum()
```

```python
# correcting column 'Paying Customer'	'Do_Not_Contact' Yes or No only

print(df['Do_Not_Contact'].unique())
```

```python
df['Do_Not_Contact'].replace('Y','Yes', inplace=True)
df['Do_Not_Contact'].replace('N','No', inplace=True)
df
```

```python
# Same for the column 'Paying customer'

df['Paying Customer'].replace('Y','Yes', inplace=True)
df['Paying Customer'].replace('N','No', inplace=True)
df
```

```python
# let's correct the column 'last_Name'
print(df[df['CustomerID'] == 1015])

# in here the "last_name" contains an _ in the end 
print("\n\n", df[df['CustomerID'] == 1015]['Last_Name'])

```

```python
df['Last_Name'] = df['Last_Name'].str.replace("_", "")
df
```

```python
# let's correct Phone_Number 

df['Phone_Number'] = df['Phone_Number'].str.replace("/", "-", regex=False)
df['Phone_Number'] = df['Phone_Number'].str.replace("\|", "-", regex=True)
df
```

```python
# drop the row where Customerid = 1019
df.drop(index=18, inplace=True)

df
```
