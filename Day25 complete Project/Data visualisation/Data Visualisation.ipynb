```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# On va importer les données nettoyées
adult = pd.read_csv('C:\\Users\\hp\\Desktop\\100 Days of data science practise\\Day25 complete Project\\Data cleaning\\clean_data.csv')
adult.head()
```

```python
adult.info()
```

```python
# On vérifie la distribution de la variable cible

adult['income'].value_counts()
```

```python
print("La classe <=50k est majaroitaire, et la classe >50k est minoritaire avec une différence de:",33988 - 11206,"valeurs.")
```

```python
plt.figure(figsize=(10,5))
sns.countplot(data= adult, x= 'income')
plt.title("Distribution de la classe 'income'", fontsize= 14)
plt.xlabel('Catégorie de revenu', fontsize=12)
plt.ylabel('Nombre d\'individus', fontsize=12)
plt.show()
```

```python
# income selon le genre
plt.figure(figsize=(8,6))
sns.histplot(data= adult, x='income', hue= 'sex', multiple='stack')
plt.title("Income selon le genre")
plt.show()
```

<p> Le nombre de femmes avec un revenu <=50k ou >50k est très minim que le nombre d'hommes

```python
plt.figure(figsize=(18,6))
sns.lineplot(data= adult, y='capital-gain', x= 'hours-per-week', hue='sex', errorbar= None)
plt.ylabel('Capital gain', fontsize= 12)
plt.xlabel('Hours per week', fontsize= 12)
plt.title('Capital gain vs Hours per week selon le genre', fontsize= 16)
plt.show()
```

```python
plt.figure(figsize=(18,6))
sns.lineplot(data= adult, y='capital-loss', x= 'hours-per-week', hue='sex', errorbar= None)
plt.ylabel('Capital loss', fontsize= 12)
plt.xlabel('Hours per week', fontsize= 12)
plt.title('Capital loss vs Hours per week selon le genre', fontsize= 16)
plt.show()
```

```python
# Matrice de corrélation
corr = adult.corr(numeric_only= True)
corr
```

```python
plt.figure(figsize=(12,6))
sns.heatmap(corr, annot= True, linewidths= 0.5, cmap= 'Blues')
plt.title('Matrice de corrélation', fontsize= 15)
plt.show()
```

Il n'y a pas de forte corrélation positive entre les variables
