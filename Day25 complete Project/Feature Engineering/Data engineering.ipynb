### 6)- La colonne 'occupation', 




```python
adult['occupation'].value_counts()
```

```python
# On va suprimer ma colonne 'occupation'
adult.drop(columns=['occupation'], inplace= True)
adult.head(2)
```

### 7)- Les colonnes 'marital-status', 'relationship'



```python
adult['marital-status'].value_counts()
```

```python
adult['relationship'].value_counts()
```

```python
# On va supprimer la colonne 'relationship' 
adult.drop(columns=['relationship'], inplace= True)
adult.head(1)
```

```python
# Pour la colonne 'marital-status' on va remplacé par deux valeurs seulement marrié ou non 
# marrié : 1 non marrié : 0
adult['married'] = adult['marital-status'].replace({
                                                                'Married-civ-spouse': 1 ,
                                                                'Never-married' : 0 ,          
                                                                'Divorced' : 0  ,                
                                                                'Separated' : 0  ,             
                                                                'Widowed' : 0 ,                  
                                                                'Married-spouse-absent' : 1 ,      
                                                                'Married-AF-spouse' : 1          
})
adult.head(2)
```

```python
# On supprime maintenant la colonne 'marital-status'

adult.drop(columns= ['marital-status'], inplace= True)
adult.head(2)
```

### 8)- Les colonnes 'race', 'native-country',
       
       


```python
adult['race'].value_counts()
```

```python
adult['native-country'].value_counts()
```

```python
# On supprime la colonne 'native-country'
adult.drop(columns= ['native-country'], inplace= True)
adult.head(2)
```

### 9)- La colonne 'income'

```python
adult['income'].value_counts()
```

```python
# On va convertir la colonne 'income' en int
adult['income'] = adult['income'].replace({
                                                                '<=50K': 0 ,
                                                                '>50K' : 1          
})
adult.head(2)
```

```python
# Pour le reste des colonnes 'workclass' 'race' on va faire le one hot encoding

adult = pd.get_dummies(adult, columns=['race', 'workclass'], drop_first= True)
adult.head()
```
