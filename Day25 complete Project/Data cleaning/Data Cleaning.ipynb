```python
# On remplace la valeur 'Outlying-US(Guam-USVI-etc)' par 'United-States' 
adult['native-country'].replace('Outlying-US(Guam-USVI-etc)', 'United-States' , inplace=True)
adult['native-country'].unique()
```

```python
# On corrige la valeur ' Trinadad&Tobago' par 'Trinadad & Tobago'
adult['native-country'].replace('Trinadad&Tobago', 'Trinadad & Tobago' , inplace=True)
adult['native-country'].unique()
```

```python
# On remplace 'Hong' qui fait référence à 'Hong kong' par 'China'
adult['native-country'].replace('Hong', 'China' , inplace=True)
adult['native-country'].unique()
```

```python
# On remplace <=50. par <=50 
adult['income'].replace('<=50K.', '<=50K', inplace=True)

# On remplace >50k. par >50k
adult['income'].replace('>50K.', '>50K', inplace=True)
```
