```python
from PIL import Image
import matplotlib.pyplot as plt

# The path of the image
path = 'faces/0805personali01.jpg'

# Open the image
try:
    img = Image.open(path)
    plt.imshow(img)
except IOError:
    pass
```

## 3- Rotating an image

```python
def rotate_image():
    try:
        # 1- We open the image
        img = Image.open(path)
        print("1- The image has been open\n ")
        
        
        # 2- We rotate the image with an angle of 90°
        img_rot = img.rotate(90) # We can add other arguments !
        print("2- The image has been rotated\n")
        
        
        # 3- We save the rotatd image
        img_rot.save("rotated_image.jpeg")
        print("3- The rotated image has been saved")
        
        
        # We can show the 2 images :
        plt.subplot(1,2,1)
        plt.imshow(img)
        plt.subplot(1,2,2)
        plt.imshow(img_rot)
    except IOError:
        pass
    
if __name__ == "__main__":
    rotate_image()
```
