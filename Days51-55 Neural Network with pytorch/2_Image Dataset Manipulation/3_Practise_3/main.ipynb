```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

# 1-

Écrire une fonction Python qui :

prend en entrée le chemin du dossier images

retourne la liste des chemins complets de toutes les images du dossier


```python
def list_Images(path):
    # le path existe ou non 
    if not os.path.exists(path):
        raise FileNotFoundError(f"Le dossier {path} n'existe pas!")
    
    list_images_path = [] # liste pour stocker les images finales
    for filename in os.listdir(path):
        filename_lower = filename.lower() # des fichiers avec .PNG et d'autres avec .png etc
        if filename_lower.endswith(('png', 'jpg', 'jfif')):
            full_path = os.path.join(path, filename) # le chemin complet !
            list_images_path.append(full_path)
    return list_images_path



image_path = 'dataset/images'
list_Images(image_path)
```

```python
# 3- Fonction qui charge TOUTES les image à partir d'un chemin

def load_images(paths):

    
    images = []
    for path in paths:
        img = load_image(path)
        images.append(img)
    return images
```

Afficher :

les 3 premières images

avec leur nom de fichier affiché au-dessus de chaque image

```python
# 4- Fonction qui affiche les 3 premiéres images avec leurs noms

def plot_images(image_path):
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Le chemin {image_path} n'existe pas ")
    
    list_images = os.listdir(image_path)
    
    fig, axes = plt.subplots(1, 3, figsize=(20,8))
    
    for idx, img in enumerate(list_images):
        if (idx < 3):
            img_pat = os.path.join(image_path, img)
            image_label = img
            img = Image.open(img_pat)
            axes[idx].imshow(img)
            axes[idx].set_title(image_label)
            axes[idx].axis('off')
    plt.tight_layout()
    plt.show()
```

```python
image_path = 'dataset/images'
plot_images(image_path)
```

```python
import matplotlib.pyplot as plt

def plot_images(images, labels=None, n=3):
    n = min(n, len(images))
    fig, axes = plt.subplots(1, n, figsize=(5*n, 5))

    if n == 1:
        axes = [axes]

    for i in range(n):
        axes[i].imshow(images[i])
        if labels is not None:
            axes[i].set_title(labels[i])
        axes[i].axis('off')

    plt.tight_layout()
    plt.show()
```
