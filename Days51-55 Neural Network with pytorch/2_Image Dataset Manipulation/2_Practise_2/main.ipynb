```python
import matplotlib.pyplot as plt
from PIL import Image
import os
```

```python
images = 'faces/'
labels = 'labels/'
```

```python
os.listdir(images), type(os.listdir(images))
```

```python
# Print an image using the name of the image

def plot_image_name(image_name):
    
    image_path = images+ image_name
    
    img = Image.open(image_path)
    plt.imshow(img)
    
```

```python
plot_image_name('0805personali01.jpg')
```

```python
# Print an image using the number

def plot_image_id(image_number: int):
    list_images = os.listdir(images)
    img = list_images[image_number]
    print("The image with the number: ", img )
    img = Image.open(images+ img)
    plt.imshow(img)
    
plot_image_id(1)
```

```python
def plot_images(path):
    images = os.listdir(path)
    rows = 2
    colons = len(images)/2
    fig, axes = plt.subplots(rows, int(colons), figsize=(18, 4))
    axes = axes.flatten()
    for idx, image in enumerate(images):
        img = Image.open(os.path.join(path, image))
        axes[idx].imshow(img)
        axes[idx].axis("off")

    plt.show()
plot_images(images)
```
