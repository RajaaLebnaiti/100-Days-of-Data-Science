```python
from PIL import Image
import os
import matplotlib.pyplot as plt
```

```python
path = 'faces/'

def plot_images(path):
    list_images = os.listdir(path)
    fig, axes = plt.subplots(2, 7, figsize=(22,6))
    axes = axes.flatten()
    for idx, img_name in enumerate (list_images):
        img = os.path.join(path, img_name)
        img = Image.open(img)
        axes[idx].imshow(img)
        axes[idx].axis("off")
    plt.tight_layout()
    plt.show()

        

plot_images(path)
```

```python
import matplotlib.pyplot as plt
from PIL import Image  
import os
import pandas as pd

faces = 'faces/'
labels = 'labels/labels.csv'

def plot_images_with_labels(images_path, labels_path):
    images = sorted(os.listdir(images_path))
    labels = pd.read_csv(labels_path)['label'].values
    
    fig, axes = plt.subplots(2, 7, figsize= (22,8))
    axes = axes.flatten()
    
    for idx, img in enumerate (images):
        axes[idx].imshow(Image.open(os.path.join(images_path, img)))
        axes[idx].set_title(labels[idx])
        axes[idx].axis('off')
    plt.tight_layout()
    plt.show()

plot_images_with_labels(faces, labels)
```
