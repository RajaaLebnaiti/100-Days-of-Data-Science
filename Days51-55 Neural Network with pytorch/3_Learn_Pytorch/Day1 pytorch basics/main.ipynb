```python
import torch

x = torch.arange(1 , 9)
print(x, "\n", x.shape)
```

```python
# reshape
x_reshaped = x.reshape(1, 8)
print(x_reshaped, "\n", x_reshaped.shape)
```

```python
# reshape
x_reshaped2 = x.reshape(2, 4)
print(x_reshaped2, "\n", x_reshaped2.shape)
```

```python
# view
x_view = x.view(1,8)
print(x_view, "\n", x_view.shape)
```

```python
#La différence entre view et reshape

import torch

x = torch.tensor([[1, 2], [3, 4]]) # Un carré 2x2
x_t = x.t()  # On transpose : les lignes deviennent colonnes

print(f"Est-ce que x_t est contigu ? {x_t.is_contiguous()}")

# Tentative avec view -> Erreur !
try:
    x_view = x_t.view(4)
except RuntimeError as e:
    print(f"\nErreur view : {e}")

# Tentative avec reshape -> Succès (car il fait une copie automatique)
x_reshape = x_t.reshape(4)
print(f"\nSuccès reshape : {x_reshape}")
```

```python
# stack
x = torch.arange(1,5)
y = torch.arange(7, 11)
x_stack = torch.stack([x,x, y, x*y] )

x, x_stack
```

```python
# squeeze
x = torch.arange(1,5)
x = x.reshape(1, 4)

x_squeeze = x.squeeze() # supprime [

print(x,"\n", x_squeeze)
```

```python
# unsqueeze
x = torch.arange(1,5)
x_unsqueeze = x.unsqueeze(dim=0) # ajoute[
print(x,"\n", x_unsqueeze)


print("-----------------------------------")
x_unsqueeze2 = x.unsqueeze(dim=1) # ajoute[
print(x_unsqueeze2)


```

```python
# permute


x_original = torch.rand(size=(224, 224, 3))

x_permuted = x_original.permute(2, 0, 1) # shifts axis 0->1, 1->2, 2->0

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```

```python
x_original = torch.rand(size=(224, 224, 3))

x_permuted = x_original.permute(0, 2, 1) 

print(f"Previous shape: {x_original.shape}")
print(f"New shape: {x_permuted.shape}")
```
