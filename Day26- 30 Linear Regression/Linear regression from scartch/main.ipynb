```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('Salary_Data.csv')

data.head()
```

```python
x_train = data['YearsExperience'].values
y_train = data['Salary'].values
```

```python
plt.scatter(x_train, y_train)
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Training Data')
plt.show()
```

<h4> 1)- MSE (Mean Squared Error)

```python
# First, we create the cost function MSE

def cost_function(x, y, w, b):
    m = len(x) 
    cost_sum = 0
    
    
    for i in range(m): 
        f = w* x[i] + b # La droite de régression = y prédite
        cost = (f - y[i])** 2
        cost_sum += cost
        
    total_cost =  (1/2*m) * cost_sum
    return total_cost
```

<h4> 2)- Gradient function

```python
# Second, we create a gradient function to compute derivatives of w and b


def gradient_function(x, y, w, b):
    m = len(x)
    dc_dw, dc_db = 0, 0
    
    for i in range(m):
        f= w* x[i] + b # droite de régression
        
        # compute the derivate of cost function for w and b
        dc_dw += (f - y[i]) * x[i]
        dc_db += (f - y[i])
        
    # retrun the two derivatives    
    dc_dw = (1/m)* dc_dw
    dc_db = (1/m)* dc_db
    
    return dc_dw, dc_db
    
```

<h4> 3)- Gradient Descent function

```python
# Second, we create a gradient function to compute derivatives of w and b

def gradient_descent_function(x, y, alpha, iteration):
    w, b = 0, 0 # intialisation des valeurs w et b
    
    for i in range (iteration):
        
        # get the derivatives of cost function
        dc_dw, dc_db = gradient_function(x, y, w, b)
        
        # update w and b
        w = w - alpha * dc_dw 
        b = b - alpha * dc_db
        print(f" Iteration {i}: Cost: {cost_function(x, y, w, b)}")
    return w , b
```

```python
# run the code
alpha = 0.01
iteration = 1000

final_w, final_b = gradient_descent_function(x_train, y_train, alpha, iteration)
```

```python
plt.scatter(x_train, y_train, color= 'blue', label= 'Data')

x_vals = np.linspace(min(x_train), max(x_train), 100)
y_vals = final_w * x_vals + final_b
plt.plot(x_vals, y_vals, color= 'red', label= 'Droite de régression')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```
