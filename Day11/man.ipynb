<h1><center> Day 11: Mean, median, variance, standard deviation

```python
import numpy as np
```

```python
data = [1, 2, 3, 4, 100]
data
```

```python
print("Mean: ", np.mean(data)) # La moyenne = 110 / 5 = 22
```

```python
print("Mediane: ", np.median(data)) # La mediane = 3
```

```python
print("Variance: ", np.var(data)) # La variance = 1/5[ (1-22)^2 + (2-22)^2 + (3-22)^2 + (4-22)^2 + (100-22)^2 ]  = 1522
```

```python
print("Ecart type: ", np.sqrt(1522))
```

```python
import statistics
data1 = [1,2, 14, 6, 2, 20, 2]
print("Mode: ", statistics.mode(data1))
```
