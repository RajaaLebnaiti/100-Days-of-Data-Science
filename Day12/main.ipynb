<h1><center> Linear Algebra

```python
import numpy as np
```

```python
# Vectors 

v1= np.array([1, 2, 3])
print(v1)
```

```python
v2= np.array([4, 5, 6])
print(v2)
```

```python
print("v1 + v2 =",v1 + v2)  #  [5, 7, 9]
```

```python
print("v1 * v2 =",v1 * v2)  #  [4, 10, 18]
```

```python
# Matrices

m1 = np.array([
                [2,1],
                [1,2]
            ])

print(m1.shape)
```

```python
m2 = np.array([
                [3,0],
                [1,2]
            ])

```

```python
print("m1 * m2 = \n", np.dot(m1, m2)) # multiplication matricielle
```

```python
# Transposée

print("la transposée de m1 est : \n", m1.transpose())
```

```python
# Inverse d'une matrice A. A^-1 = I

m1Inverse = np.linalg.inv(m1)
print(m1Inverse)
```

```python
# To make sure 

print(np.dot(m1, m1Inverse))  # ----> Doit donner la matrice identitée
```

```python
# Determinant

print(np.linalg.det(m1))  # --> determinant d'une matrice Si det(A) = 0 alors la matrice est non inversible
```

<h3> PCA (Principal Component Analysis)

<p>It helps you to reduce the number of features in a dataset while keeping the most important information.

<p><b>PCA uses linear algebra to transform data into new features called principal components. It finds these by calculating eigenvectors (directions) and eigenvalues (importance) from the covariance matrix. PCA selects the top components with the highest eigenvalues and projects the data onto them simplify the dataset.

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
```

```python
# Creating a small data set 

data = {
    
    'Height': [170, 165, 180, 175, 160, 172, 168, 177, 162, 158],
    'Weight': [65, 59, 75, 68, 55, 70, 62, 74, 58, 54],
    'Age': [30, 25, 35, 28, 22, 32, 27, 33, 24, 21],
    'Gender': [1, 0, 1, 1, 0, 1, 0, 1, 0, 0]  # 1 = Male, 0 = Female

}

data
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
# division des données en features et target
X = df.drop('Gender', axis=1)
y = df['Gender']

print(X, "\n\n", y)
```

```python
# calcule de mean et standard deviation

Xmean = X.mean()
Xvariance = X.var()
Xdeviation = np.sqrt(Xvariance)

print("\nThe mean is : \n", X.mean())
print("\nThe variance is : \n", Xvariance)
print("\nThe standard deviation is : \n", Xdeviation)
```

```python
# standarisation des données
scaler = StandardScaler()
X_scaled= scaler.fit_transform(X)

print(X_scaled)
```

```python
# Applying PCA

pca = PCA(n_components= 2)
X_pca = pca.fit_transform(X_scaled)


print(X_pca)
```

```python
# Train the model

X_train, X_test, y_train, y_test = train_test_split(X_pca, y, test_size= 0.3, random_state= 42)

print(len(X_train), len(y_train), len(X_test), len(y_test))
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)

print(y_pred, "\n\n", y_test)
```

```python
model.score(X_test, y_pred)
```

```python
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
