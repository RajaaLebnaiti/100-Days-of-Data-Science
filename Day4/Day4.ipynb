# Note: 
Closing a file will free up the resources that are tied to the file. Hence, it is a good programming practice to always close the file.

```python
# Better way to open a file:
with open("notes.txt", "r") as file:
    read_contents = file.read()
    print(read_contents)
```

Here, with...open automatically closes the file, so we don't have to use the close() function.

```python
# Best practise to use try finally

try:
    file = open("notes.txt", "r")
    read_contents = file.read()
    print(read_contents)
finally:
    file.close()
```

# Handling CSV files

```python
import csv
with open ("people.csv", "w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["Name", "Age"])
    writer.writerow(["Rajaa", 23])
    writer.writerow(["Omar", 25])
    
with open("people.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

# Handling Json files

```python
import json

# dictionnary to store data
data = {
    "name": "Rajaa",
    "age": 23,
    "skills": ["Python", "Data Science"]
}

# write in json file
with open("person.json", "w") as file:
    json.dump(data, file)
    
    
# read a json file
with open("person.json", "r") as file:
    loaded = json.load(file)
    print(loaded)
```

```python
# Roll a dice game

rand_num = randint(1, 6)
num = 0

while(rand_num != num):
    num = int(input("Enter a number from 1 to 6"))
    try:
        if num == rand_num:
            print("You win!!")
        else :
            print("Try again :( ")
    except ValueError:
        print("Enter a valid number")
```
