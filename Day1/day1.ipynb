```python
# Calculator

def calculate(a,b, operation):
    try:
        if operation == "add":
            return a + b
        elif operation == "substract":
            return a - b
        elif operation == "multiply":
            return a * b
        elif operation == "divide":
            return a / b
        else:
            return "Operateur saisit est incorrect. Réssayer!"
    except ValueError:
        print("Wrong operation!")
        
        
calculate(1,2,"divide")
    
```
