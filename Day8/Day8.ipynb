# Cleaning data

# empty cells

## 1 drop null values

```python
data.isnull().sum()
```

```python
new_data = data.dropna()
new_data.isnull().sum()
```

By default, the dropna() method returns a new DataFrame, and will not change the original.

If you want to change the original DataFrame, use the inplace = True argument

```python
print("before: \n", data.isnull().sum())
data.dropna(inplace=True)

print("\n\n\nafter: \n", data.isnull().sum())
```

## 2 replace null values

```python
# we can also replace 
data = pd.read_csv("data.csv")
data.fillna({"Calories":130}, inplace=True) # we repleace empty values with 130 ONLYYYYYY IN column Calories
```

```python
# we can also replace using mean(), median(), max()

data = pd.read_csv("data.csv")

mean_calories = data['Calories'].mean()
mean_calories
```

```python
data.fillna({"Calories": mean_calories}, inplace= True)
```

```python
# same thing for mean
```

```python
# for mode
df = pd.read_csv('data.csv')

x = df["Calories"].mode()
print(x)
print("\n\n", x[0])
```

```python
df.fillna({"Calories": x}, inplace=True)
# Mode = the value that appears most frequently.

```

In our Data Frame, we have two cells with the wrong format. Check out row 22 and 26, the 'Date' column should be a string that represents a date:

```python
data['Date'] 
```

```python
pd.options.display.max_rows = 1000
```

```python
# in row 26 we have  26/12/2020 but it should be '2020/12/26'

# to correct this we use
df = pd.read_csv('data1.csv')

print(df)
```

```python
df.dropna(subset=['Date'], inplace = True) # remove the null the date
```

```python
df
```

```python
df['Date'] = df['Date'].str.replace("'", "")
df['Date'] = pd.to_datetime(df['Date'], format='mixed')

print(df.to_string())
```
