<h1><center>Probability Fundamentals

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, uniform, norm
```

```python
data_uniform = uniform.rvs(size= 10000, loc=0, scale=10)  # 1000 valeurs entre 0 et 10
data_uniform
```

```python
# -------- Distribution Binomiale --------
n, p = 10, 0.5   # 10 essais, probabilité de succès = 0.5
x_binom = np.arange(0, n+1)
data_binom = binom.pmf(x_binom, n, p)
data_binom
```

```python
# -------- Distribution Normale --------
mu, sigma = 0, 1   # moyenne = 0, écart-type = 1
x_norm = np.linspace(-4, 4, 1000)
data_norm = norm.pdf(x_norm, mu, sigma)
data_norm
```

```python
# -------- Plotting --------
plt.figure(figsize=(15,4))

# Uniforme
plt.subplot(1,3,1)
plt.hist(data_uniform, bins=30, color="skyblue", edgecolor="black", density=True)
plt.title("Distribution Uniforme (0-10)")

# Binomiale
plt.subplot(1,3,2)
plt.bar(x_binom, data_binom, color="orange", edgecolor="black")
plt.title("Distribution Binomiale (n=10, p=0.5)")

# Normale
plt.subplot(1,3,3)
plt.plot(x_norm, data_norm, color="red")
plt.title("Distribution Normale (μ=0, σ=1)")

plt.tight_layout()
plt.show()
```
