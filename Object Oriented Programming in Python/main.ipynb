### Attributes and Methods

```python
class Customer:
    
    # method 1: 
    def identify(name):
        print("I am customer " + name)
```

```python
customer1 = Customer()
```

```python
customer1.identify("Rajaa")
```

En Python, lorsqu'une méthode est appelée à partir d'une instance : <i>customer1.identify("Rajaa")</i>

Python le traduit en interne par :

<i>Customer.identify(customer1, "Rajaa") </i>

L'objet lui-même (customer1) est envoyé comme premier argument à la méthode. C'est ce qu'on appelle conventionnellement self.

```python
# Il faut ajouter l'argument self!
class Customer:
    
    def identify(self, name):
        print("I am customer " + name)
        

customer2 = Customer()
customer2.identify("Rajaa")
```

```python
# To create attributes

class Customer:
    
    def set_name(self, new_name):
        self.name = new_name
        
    def identify(self):
        print("I am customer " + self.name)
        
        
        
        
        
        
           
cust = Customer()
cust.set_name("Hello world")
```

### _ _ init _ _() constructor

```python
class Customer:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        
        
cust1 = Customer("Rajaa", 24)

print("Customer name : "+ cust1.name)
print("Customer age : ", cust1.age)
```

```python
class Customer:
    def __init__(self, name, age, salary = 0):
        self.name = name
        self.age = age
        self.salary = salary
        
cust1 = Customer("Rajaa", 24)
cust2 = Customer("John", 30, 1000)


print("Customer name : "+ cust1.name, "Customer age : ", cust1.age, "Customer salary :", cust1.salary)
print("Customer name : "+ cust2.name, "Customer age : ", cust2.age, "Customer salary :", cust2.salary)
```

```python
# Write the class Point as outlined in the instructions
import numpy as np
class Point:
    def __init__(self, x= 0, y= 0):
        self.x = x
        self.y = y

    def distance_to_origin(self):
        return np.sqrt((self.x * self.x) + (self.y * self.y))

    def reflect(self, axis):
        if axis == "x":
            self.y = self.y * (-1) 
        if axis == "y" :
            self.x = self.x * (-1)
        else:
            print("Error!")      
            
            
            
            
            
            
            
            
pt = Point(x=3.0)
pt.reflect("y")
print((pt.x, pt.y))
pt.y = 4.0
print(pt.distance_to_origin())
```
