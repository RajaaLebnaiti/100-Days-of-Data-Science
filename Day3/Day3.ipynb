# Explain *args and **kwargs

```python
def add(a,b):
    return a + b

add(1, 2)

# Now if we want to add another argument like:
# add(1, 2, 3) # --> This funtion is no longer working!
```

```python
# That' why we use * args , we put all our arguments into a tuple

def add(*args):
    total = 0
    for arg in args:
        total += arg
    return total


add(1,3) 

# Now we can add as many argument as we want without changin the function
add(1, 2, 3)
add(1, 2, 3, 4, 5)
```

```python
# Another example:

def display_name(*args):
    for arg in args:
        print(arg, end=" ")
        
display_name("Rajaa", "Lebnaiti")
```

```python
def print_adress(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
print_adress(street="123 Faka street" , 
            city= "Detroite", 
            state= "Michigan" , 
            zip_code="54321" )
```

```python
def shipping_label(*args, **kwargs):
    for arg in args :
        print(arg, end=" ")
    print()
    for key, value in kwargs.items():
        print(f"{key} : {value} ")
```

```python
shipping_label("Dr.", "XXX", "YYY", "III", # args
               
                street="123 Faka street" ,  # now we define kwargs
                city= "Detroite", 
                state= "Michigan" , 
                zip_code="54321")
```

# Lambda function

```python
# Lambda function

add_1 = lambda x,y,z : x + 1 + y - z
add_1(3,2,1)
```

```python
def myfunc(n):
    return lambda a : a * n

value = myfunc(2) 
value(12)
```

```python
def multiply(num):
    return lambda a : a * num

mydoubler = multiply(2)
mytripler = multiply(3)

print(mydoubler(11))
print(mytripler(11))
```

```python
s2 = lambda string : string.upper()

s1 = "Hello World"
s2(s1)


```

```python
n = lambda x : "Positive" if x > 0   else "Negative" if x < 0     else "Zero"

print(n(2))
print(n(-2))
print(n(0))
```

```python
li = [lambda arg = x :   arg * 10 for x in range(1,5)]
for i in li:
    print(i())
```

```python
# check if a  number is even or not 

check = lambda x: "Even" if x%2 == 0   else "Odd"

print(check(2))
print(check(3))
```

```python
# Perform addition and multiplication in a single line
calc = lambda x,y: (x+y, x*y, x-y, x/y)

calc(2,3)
```

```python
# Example: Double each number in a list
a = [1, 2, 3, 4]
b = map(lambda x: x * 2, a)
print(list(b))
```

```python
from functools import reduce

# Example: Find the product of all numbers in a list
a = [1, 2, 3, 4]
b = reduce(lambda x, y: x * y, a)
print(b)
```

```python
# map(function, collection) = applies a given function to all items in a collection

celsius_temps = [0.0 , 10.0, 20.0, 30.0, 40.0, 50.0]

# we went to convert from celsius to ferhneit

# Fisrt we define a function 
def c_to_f(temp):
    return (temp* 9/5) + 32
    
    
# fahrenheit_temp = c_to_f(celsius_temps) # gaves an error 

faherenheit_temp = map(c_to_f,  celsius_temps)
for i in faherenheit_temp:
    print(i)
```
