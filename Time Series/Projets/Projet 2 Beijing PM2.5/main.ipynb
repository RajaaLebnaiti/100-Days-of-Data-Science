Le point de rosée est la température à laquelle l'air doit être refroidi, à pression constante, pour que la vapeur d'eau qu'il contient se transforme en eau liquide (condensation).
-  Un point de rosée inférieur à 15 °C (50-60 °F) est confortable.
-  Entre 16 °C et 18 °C, l'air commence à paraître lourd ou humide.
-  Au-dessus de 20 °C, l'atmosphère devient très lourde et moite
