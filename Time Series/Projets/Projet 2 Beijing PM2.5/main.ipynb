Le point de rosée est la température à laquelle l'air doit être refroidi, à pression constante, pour que la vapeur d'eau qu'il contient se transforme en eau liquide (condensation).
-  Un point de rosée inférieur à 15 °C (50-60 °F) est confortable.
-  Entre 16 °C et 18 °C, l'air commence à paraître lourd ou humide.
-  Au-dessus de 20 °C, l'atmosphère devient très lourde et moite

## 3. Décomposition de la série temporelle

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomposition = seasonal_decompose(df['pm2.5'], model= "additive", period= 365)

fig = decomposition.plot()
fig.set_size_inches(32, 8)
plt.suptitle("Décomposition additive de la série temporelle période= 365")
plt.show()
```

## 4. Vérifer la stationarité de la série

```python
# Méthode 1:  méthode statistique Moyenne et variance constante

window_size = 8760 # 365* 24 = 8760 HEURES

rolling_mean = df['pm2.5'].rolling(window_size).mean()
rolling_var = df['pm2.5'].rolling(window_size).var()

# visualisation
plt.figure(figsize=(22, 6))

plt.subplot(2, 1, 1)
plt.plot(df['pm2.5'], label= "Série temporelle brute", color= 'blue')
plt.plot(rolling_mean, label= "Moyenne glissante", color= "red")
plt.legend()
plt.title("Visualisation de la moyenne glissante et la série brute")

plt.subplot(2, 1, 2)
plt.plot(rolling_var, label= "Variance glissante", color= "green")
plt.legend()
plt.title("Visualisation de la mvariance glissante")

plt.tight_layout()
plt.grid(True)
plt.show()
```

```python
# Méthode 2: les testes statistiques

from statsmodels.tsa.stattools import adfuller

result_brut = adfuller(df['pm2.5'], result_object=False)

p_value = result_brut[1]

print(f"p_value = {p_value}.")


if p_value > 0.05:
    print("La série est non stationnaire.")
if p_value < 0.05:
    print("La série est stationnaire.")
```
